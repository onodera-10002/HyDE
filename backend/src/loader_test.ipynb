```python
import config
from docling.datamodel.document import TableItem
import neologdn
import pandas as pd
# ダミーDoclingDocumentの作成

from docling.datamodel.document import DoclingDocument, TextItem, TableItem
from docling.datamodel.base_models import DocItemLabel
from docling.document_converter import DocumentConverter

converter = DocumentConverter()
WEB_URL = "https://www.jstage.jst.go.jp/article/taxa/43/0/43_30/_pdf"
doclingdocument = converter.convert(WEB_URL).document

def _normalization(docs, WEB_URL=WEB_URL):
        document = docs
        print(document)
        clean_document = []
        for item, _ in document.iterate_items():
            content = ""
            p = item.prov[0] if item.prov else None
            page_no = p.page_no
            if item.label in config.NORMALIZE_LABELS:
                if hasattr(item, "text") and item.text:
                    item.text = neologdn.normalize(item.text)
                    content = item.text
                    

            elif item.label == "table":
                if isinstance(item, TableItem):
                    df = item.export_to_dataframe()
                    df_cleaned = df.map(lambda x: neologdn.normalize(str(x)) if pd.notnull(x) else "")
                    content = df_cleaned.to_markdown(index=False)
            else:
                content = item

            clean_document.append({
                "content": content,
                "page_no": page_no,
                "source": WEB_URL
            })
        return clean_document


result = _normalization(doclingdocument)
for entry in result:
    print(f"Page {entry['page_no']}:\n{entry['content']}\n{entry['source']}\n")
```
